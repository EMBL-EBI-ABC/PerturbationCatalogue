```python
from pathlib import Path

from curation_tools.llm_curation.publication_text import (
    retrieve_pub_full_text,
    pdf_to_md,
)
from curation_tools.llm_curation.llm_curation_schema import CurationSchema
from curation_tools.llm_curation.xml_parser import xml_to_md
import curation_tools.llm_curation.metadata_extraction as metadata_extraction_module
import curation_tools.llm_curation.mavedb.metadata_extraction_runner as mavedb_runner_module

extract_metadata_from_publication = metadata_extraction_module.extract_metadata_from_publication

MAVEDB_EXTRACTION_KWARGS = {
    "output_dir": mavedb_runner_module.EXTRACTED_METADATA_OUTPUT_DIR,
    "log_file": mavedb_runner_module.METADATA_EXTRACTION_LOG_FILE,
    "prompt_template_file": mavedb_runner_module.PROMPT_TEMPLATE_FILE,
    "prompt_context_builder": mavedb_runner_module.prompt_context_builder,
    "prompt_context_formatter": mavedb_runner_module.format_supplementary_mavedb_metadata,
    "context_output_suffix_builder": mavedb_runner_module.context_output_suffix_builder,
    "output_metadata_builder": mavedb_runner_module.output_metadata_builder,
}

DOWNLOADS_DIR = Path('~/Downloads').expanduser()
```

```python
filepath = retrieve_pub_full_text(
    doi='10.1016/j.ajhg.2018.03.018',
    output_dir=DOWNLOADS_DIR,
)

if filepath.suffix == '.pdf':
    publication_full_text_path = pdf_to_md(
        filepath_pdf=filepath,
        output_dir=DOWNLOADS_DIR,
        remove_references=True,
    )
elif filepath.suffix == '.xml':
    publication_full_text_path = xml_to_md(
        filepath_xml=filepath,
        output_dir=DOWNLOADS_DIR,
        remove_references=True,
    )

extract_metadata_from_publication(
    publication_full_text_path=publication_full_text_path,
    model_name='google/gemini-3.5-flash',
    extraction_schema=CurationSchema,
    overwrite=True,
    **MAVEDB_EXTRACTION_KWARGS,
)
```
